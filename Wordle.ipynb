```python
# Make a function to take a guess and a answer

# Correct letter + correct position (green)

# Letter isn't available (grey)

def wordle_res(guess, answer):
    for char1, char2 in zip(guess, answer):
        if char1 == char2:
            print('🟩', end='')
        else:
            print('⬛', end='')

g = 'slate'
ans = 'crane'

wordle_res(g, ans)
```

```python
# Correct letter + wrong position (yellow only)

def wordle_res_yellow(guess, answer):

    result = ['⬛'] * 5
    answer_chars = list(answer)

    # Find green letters
    for i in range(5):
        if guess[i] == answer[i]:
            result[i] = '🟩'
            answer_chars[i] = None  # Use up the char in list
    
    # Find yellow letters
    for j in range(5):
        if result[j] == '🟩':
            continue

        if guess[j] in answer_chars:
            result[j] = '🟨'
            index = answer_chars.index(guess[j])
            answer_chars[index] = None

    print(''.join(result))

g = 'apple'
ans = 'plate'

wordle_res_yellow(g, ans)

#---------------------------------------------------------------------#

# FULL WORKING FUNCTION FOR DUPLICATES AND GREENS, YELLOWS, GREYS

#---------------------------------------------------------------------#
```
